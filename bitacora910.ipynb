1.Comence creando una carpeta para el nuevo proyecto

`mkdir pro2`

##2.**Revisar la calidad**.
Cheque la calidad de mi secuencia con fastqc, primero cree otra carpeta para añadir mis resulatdos

`mkdir fastqc`

`fastqc /projects/share/CURSO/Ch_strain_1.fastq /projects/share/CURSO/Ch_strain_2.fastq -o fastqc`

obtuve archivos .html y .zip de los archivos Ch_1 y Ch_2. Descargue los html para visualizarlos en la computadora.

`scp -P 10022 usuario@iP:/home/usuario/pro2/fastqc/*.html .`

Podría descomprimir los archivos .zip con el codigo:

`unzip *.zip`

##3.**Filtrado**.
Para realizar el filtrado y la visualización de la calidad use fastp; creo una nueva carpeta para colocar los resultados

`mkdir fastp`

`fastp -i /projects/share/CURSO/Ch_strain_1.fastq -I /projects/share/CURSO/Ch_strain_2.fastq -o fastp/Ch_s_1.filt.fastq -O fastp/Ch_s_2.filt.fastq`

Movi los archivos fastp.html y fastp.json a mi carpeta "fastp". Descargue el archivo .html para visualizarl en la computadora.

`scp -P 10022 usuario@iP:/home/usuario/pro2/fastp/fastp.html .`

##4.**Ensamble de NOVO**
Para realizar el ensamble del genoma ocupe Spades dentro de el ambiente de **conda**

`conda activate spades`

`spades.py -k 33,37,41 -t 2 -m 7 --pe1-1 /projects/share/CURSO/Ch_strain_1.fastq --pe1-2 /projects/share/CURSO/Ch_strain_2.fastq -o Spades`

`conda deactivate`

Obtuve varios archivos, entre ellos el de scaffolds.fasta y algunas graficas en archivos .gfa. Visualice el archivo .gfa en Bandage (interfaz gráfico)

![Imagen5](https://drive.google.com/uc?export=view&id=16HwFYTgwz2anu4om51iLD_HgsFnh0MCX)

##5.**Analisis de calidad**
Para revisar y evaluar el ensamble utilice quast
`quast --split-scaffolds -t 1 Spades/scaffolds.fasta`

Baje el archivo "report.html" a la computadora para ver los resultados.

`scp -P 10022 usuario@iP:/home/usuario/pro2/quast_results/latest/report.html .`
